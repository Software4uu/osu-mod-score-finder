# osu! Mod Score Finder Beta

Lokale Beta fuer eine osu!-Score-Webseite. API-Daten, lokale Pfade und Score-Daten bleiben auf dem eigenen PC.

## Setup

1. `setup-beta.bat` starten.
2. Fehlende Installationspunkte auswaehlen. Vorhandene Punkte sind ausgegraut.
3. Client ID, Client Secret, osu! stable Ordner und osu!lazer Ordner eintragen.
4. Speichern.
5. Danach `start-beta.bat` starten.

## osu! API

OAuth-App erstellen unter: https://osu.ppy.sh/home/account/edit

Callback URL:

```text
http://localhost:5173/callback
```

## Nicht zu GitHub hochladen

- `.env`
- `data/`
- `node_modules/`
- `*.log`
- lokale osu!-Datenbanken oder Replay-Dateien

## Updates

Die App prueft GitHub auf eine neuere Version und zeigt den Status oben rechts an.

- `Aktuell`: die lokale Version passt zur neuesten bekannten GitHub-Version.
- `Update verfuegbar`: der lokale Updater kann gestartet werden.
- Git-Clones werden mit `git pull --ff-only origin main` aktualisiert.
- GitHub-ZIP-Downloads werden ueber das neueste ZIP aktualisiert.
- `.env`, `data/`, `node_modules/`, Logs und lokale osu!-Ordner bleiben erhalten.
- Nach dem Update die laufende App schliessen und `start-beta.bat` erneut starten.

## Wichtige Filter

- **Scan-Seiten:** wie viele Recent-Seiten aus der osu!api neu gesammelt werden.
- **Anzeige-Limit:** wie viele Ergebnisse sichtbar sind.
- **Bester Try:** entscheidet, welcher Try pro Map-Difficulty behalten wird.
- **Improvement:** vergleicht letzten Try, letzte Stunde oder heutigen Tag.
- **Live-Scanner:** prueft nach einer Suche automatisch auf neu gespeicherte lokale Scores.
- **Kalender:** zeigt ein Monatsraster, gespielte Tage und alle gespeicherten Scores fuer einen ausgewaehlten Tag.
- **Kalender-PP-Range:** filtert den Kalender z. B. auf 100pp bis 250pp und markiert das Monats-Top-Play orange.
- **Details:** zeigt alle gespeicherten Tries auf derselben Map-Difficulty.
