# osu! Mod Score Finder Beta

Lokale Beta fuer eine osu!-Score-Webseite. API-Daten, lokale Pfade und Score-Daten bleiben auf dem eigenen PC.

## Setup

1. `setup-beta.bat` starten.
2. Fehlende Installationspunkte auswaehlen. Vorhandene Punkte sind ausgegraut.
3. Client ID, Client Secret, osu! stable Ordner und osu!lazer Ordner eintragen.
4. Speichern.
5. Danach `start-beta.bat` starten.

## osu! API

OAuth-App erstellen unter: https://osu.ppy.sh/home/account/edit

Callback URL:

```text
http://localhost:5173/callback
```

## Wichtige Filter

- **Scan-Seiten:** wie viele Recent-Seiten aus der osu!api neu gesammelt werden.
- **Anzeige-Limit:** wie viele Ergebnisse sichtbar sind.
- **Bester Try:** entscheidet, welcher Try pro Map-Difficulty behalten wird.
- **Improvement:** vergleicht letzten Try, letzte Stunde oder heutigen Tag.
